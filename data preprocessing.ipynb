```python
year_cols = [str(year) for year in range(2000, 2024)]
```

```python

df_isp_subscribers['Total Subscribers'] = df_isp_subscribers[year_cols].sum(axis=1)
```

```python

top_isp = df_isp_subscribers.sort_values(by='Total Subscribers', ascending=False).head(10)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='Total Subscribers', y='ISP Name', data=top_isp, palette='coolwarm')
plt.title('Top 10 ISPs by Total Subscribers (2000–2023)')
plt.xlabel('Total Subscribers')
plt.ylabel('ISP Name')
plt.grid(True, axis='x')
plt.tight_layout()
plt.show()
```

```python
top_5_isp_names = top_isp['ISP Name'].head(5)
```

```python
# Filter dataset
df_top5 = df_isp_subscribers[df_isp_subscribers['ISP Name'].isin(top_5_isp_names)]
```

```python
plt.figure(figsize=(12,6))
for _, row in df_top5.iterrows():
    plt.plot(year_cols, row[year_cols], marker='o', label=row['ISP Name'])

plt.title('Subscriber Trends (2000–2023) for Top 5 ISPs')
plt.xlabel('Year')
plt.ylabel('Subscribers')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
print(df_states_usage.columns.tolist())
df_states_usage.head()
```

```python
df_states_usage.columns = df_states_usage.columns.str.strip()
```

```python
year_cols_states = [str(year) for year in range(2000, 2024)]
```

```python
df_states_usage[year_cols_states] = df_states_usage[year_cols_states].apply(pd.to_numeric, errors='coerce')
```

```python
df_states_usage['Avg Usage (%)'] = df_states_usage[year_cols_states].mean(axis=1)
df_states_usage['Total Usage'] = df_states_usage[year_cols_states].sum(axis=1)
```

```python
top_avg_states = df_states_usage.sort_values('Avg Usage (%)', ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x='Avg Usage (%)', y='State Name', data=top_avg_states, palette='cubehelix')
plt.title('Top 10 States by Average Internet Usage (2000–2023)')
plt.xlabel('Average Internet Usage (%)')
plt.ylabel('State')
plt.grid(True, axis='x')
plt.tight_layout()
plt.show()
```

```python
top_5_state_names = top_avg_states['State Name'].head(5)
df_top5_states = df_states_usage[df_states_usage['State Name'].isin(top_5_state_names)]

plt.figure(figsize=(12,6))
for _, row in df_top5_states.iterrows():
    plt.plot(year_cols_states, row[year_cols_states], marker='o', label=row['State Name'])

plt.title('Internet Usage Trends (2000–2023) for Top 5 States')
plt.xlabel('Year')
plt.ylabel('Internet Usage (%)')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

heatmap_data = df_states_usage.set_index('State Name')[year_cols_states]

plt.figure(figsize=(16,14))
sns.heatmap(heatmap_data, cmap='YlGnBu', linewidths=0.3, linecolor='gray')
plt.title('Internet Usage Heatmap by State (2000–2023)')
plt.xlabel('Year')
plt.ylabel('State')
plt.tight_layout()
plt.show()
```

```python
selected_states = ['Maharashtra', 'Kerala', 'Bihar', 'Punjab', 'Tamil Nadu']
df_selected = df_states_usage[df_states_usage['State Name'].isin(selected_states)]

plt.figure(figsize=(12,6))
for _, row in df_selected.iterrows():
    plt.plot(year_cols_states, row[year_cols_states], marker='o', label=row['State Name'])

plt.title('Internet Usage Trends (2000–2023) - Selected States')
plt.xlabel('Year')
plt.ylabel('Usage (%)')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

melted = df_states_usage.melt(id_vars=['State Name'], value_vars=year_cols_states,
                              var_name='Year', value_name='Internet Usage (%)')

plt.figure(figsize=(14,6))
sns.boxplot(data=melted, x='Year', y='Internet Usage (%)', palette='Spectral')
plt.title('Distribution of Internet Usage Across States by Year')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14,6))
sns.violinplot(data=melted, x='Year', y='Internet Usage (%)', palette='coolwarm', inner='quartile')
plt.title('Internet Usage Spread Across States (2000–2023)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

df_isp_subscribers.columns = df_isp_subscribers.columns.str.strip()
df_states_usage.columns = df_states_usage.columns.str.strip()

years = [str(year) for year in range(2000, 2024)]

isp_total = df_isp_subscribers[years].sum()
state_total = df_states_usage[years].sum()
```

```python
plt.figure(figsize=(12,6))
plt.plot(years, isp_total, label='ISPs - Total Subscribers', marker='o')
plt.plot(years, state_total, label='States - Total Usage %', marker='s')
plt.title('Internet Growth: ISPs vs States (2000–2023)')
plt.xlabel('Year')
plt.ylabel('Total Value')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
isp_normalized = (isp_total - isp_total.min()) / (isp_total.max() - isp_total.min())
state_normalized = (state_total - state_total.min()) / (state_total.max() - state_total.min())

plt.figure(figsize=(12,6))
plt.plot(years, isp_normalized, label='ISPs (normalized)', marker='o')
plt.plot(years, state_normalized, label='States (normalized)', marker='s')
plt.title('Normalized Growth: ISPs vs States (2000–2023)')
plt.xlabel('Year')
plt.ylabel('Normalized Value (0-1)')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
isp_growth = isp_total.diff().fillna(0)
state_growth = state_total.diff().fillna(0)

growth_df = pd.DataFrame({
    'Year': years,
    'ISP Growth': isp_growth.values,
    'State Growth': state_growth.values
})

plt.figure(figsize=(14,6))
sns.barplot(data=growth_df.melt(id_vars='Year', var_name='Source', value_name='Growth'),
            x='Year', y='Growth', hue='Source', palette='Set2')
plt.title('Year-wise Internet Growth: ISPs vs States')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
top5_states = df_states_usage[['State Name'] + years].copy()
top5_states['Total'] = top5_states[years].sum(axis=1)
top5_states = top5_states.sort_values('Total', ascending=False).head(5)

top5_isps = df_isp_subscribers[['ISP Name'] + years].copy()
top5_isps['Total'] = top5_isps[years].sum(axis=1)
top5_isps = top5_isps.sort_values('Total', ascending=False).head(5)

# Plotting comparison
plt.figure(figsize=(14,6))
for _, row in top5_states.iterrows():
    plt.plot(years, row[years], label=f"{row['State Name']} (State)", linestyle='--')
for _, row in top5_isps.iterrows():
    plt.plot(years, row[years], label=f"{row['ISP Name']} (ISP)", linestyle='-.')

plt.title('Top 5 States vs Top 5 ISPs (2000–2023)')
plt.xlabel('Year')
plt.ylabel('Internet Usage/Subscribers')
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
